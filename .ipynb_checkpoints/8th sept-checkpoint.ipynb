```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("./iris.csv")
df
```

```python
print("\n First 5 rows:")
df.head(5)
```

```python
print("\n Dataset Information:")
print(df.info)
```

```python
print("\n Dataset Shape:")
print(df.shape)
```

```python
print("\n Stastical Summary:")
df.describe()
```

```python
print("\n Duplicates:")
print(df.duplicated().sum())
```

```python
#HISTOGRAM
df.hist(figsize=(10,8),bins=15)
plt.suptitle("Feature Distribution")
plt.show()
```

```python
#BOX PLOT
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.title("Feature Distriburation and Outliers")
plt.xticks(rotation=30)
plt.show()
```

```python
#PAIRPLOT
sns.pairplot(df,hue="species")
plt.show()
```

```python
#CORRELATION HEATMAP
numeric_df=df.select_dtypes(include="number")
plt.figure(figsize=(8,6))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)
```

```python
#COUNTPLOT
plt.figure(figsize=(8,2))
sns.countplot(
    data=df,
    x="species"
)

plt.title("Number of samples in iris species")
plt.xlabel("species")
plt.ylabel("Number of species")
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="sepal_width",
    hue="species",
    s=80
)
plt.title("sepal length vs sepal width")
plt.xlabel("sepal length(cm)")
plt.ylabel("sepal width(cm)")
plt.show()
```
