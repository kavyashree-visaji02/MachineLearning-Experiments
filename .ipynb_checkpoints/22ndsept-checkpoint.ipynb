```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
        
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error
```

```python
df=pd.read_csv('./Salary data (2).csv')
df.head()
```

```python
plt.scatter(data=df,x="YearsExperience",y="Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

```python
X=pd.DataFrame(df['YearsExperience'])
y=df['Salary']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2,random_state=1
)
```

```python
X.shape
```

```python
y.shape
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
predict=model.predict(X_test)
```

```python
predict
```

```python
y_test
```

```python
score=r2_score(predict,y_test)*100
score
```

```python
mse=mean_squared_error(y_test,predict)
mse
```

```python
plt.scatter(X_test,y_test,color='red',label='test values')
plt.scatter(X_test,predict,color='blue',label='prediction values')
plt.title("Salary data test values vs predicted values")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```
